### Write a program to find the current date of a car.

```python
import datetime
```

```python
class car:

    def __init__(self,mileage,year,make,model):
        self.mileage = mileage
        self.year =year
        self.make = make
        self.model = model

    def age(self,current_year):
        age = current_year-self.year
        return age
    
    def __str__(self):
        return "This is my car class that I have craeted"
    

```
