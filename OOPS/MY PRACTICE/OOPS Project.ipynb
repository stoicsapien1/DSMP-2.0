#### Count number of upper case,lower case,vowels,digits,space,and special symbols from a given string

```python
class String:

    def __init__(self,msg):
        self.uppercase =0
        self.lowercase =0
        self.vowels =0
        self.digits =0
        self.space =0
        self.specialsymbols =0
        self.msg =msg

    def count_uppercase(self):
        count =0
        for i in self.msg:
            if i.isupper():
                count+=1
        return count

    def count_lowerrcase(self):
        count =0
        for i in self.msg:
            if i.isupper():
                count+=1
        return count
    


    


    


```
