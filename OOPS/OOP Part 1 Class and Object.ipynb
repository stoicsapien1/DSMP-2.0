```python
class Atm:

    def __init__(self):
        self.pin=""
        self.balance = 0
        self.menu()


    def menu(self):

        inu = input("""Hi,How can I help you?
                    
                    1. Create Pin?
                    2. Change pin?
                    3. Check Balance?
                    4. Withdraw cash
                    5. exit""")
        if inu=="1":
            self.create_pin()

            

        elif inu=="2":
            self.change_pin()

            # change pin

        elif inu =="3":
            self.check_balance()

            # check balance
        
    
        elif inu == "4":
            pass

            self.withdraw()

        else:
            exit()

    def create_pin(self):

        user_pin= input("Enter your pin")

        self.pin = user_pin

        user_balance = int(input("Enter the balance"))

        self.balance = user_balance

        print("pin created successfully")
        self.menu() 

    def change_pin(self):

        

        old_pin = input("Enter your old pin?")

        if old_pin==self.pin:
            new_pin= input("Enter your new pin?")
            self.pin=new_pin
            print("Pin changed successfully")
            self.menu()
        else:
            print("Enter right old pin")
            self.menu()

    def check_balance(self):
        user_pin1=input("Enter Pin")
        if user_pin1==self.pin:
            print("Your balance is",self.balance)
        else:
            print("Enter right pin")
        self.menu()

    

    def withdraw(self):
        user_pin=input("Enter your pin?")

        if user_pin==self.pin:
            bal_withdraw = int(input("How much you want to withdraw?"))
            if bal_withdraw<=self.balance:
                print("Money withdrwan successfully")
                self.balance=self.balance-bal_withdraw
                print("Available balance",self.balance)
                
                
            
            else:
                print("payment declined",bal_withdraw,"is greater than balance present")
                
        else:
            print("Enter the right pin")

        self.menu()
        
```
