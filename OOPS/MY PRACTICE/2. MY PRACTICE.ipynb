```python
class Rectangle:

    def __init__(self,l,b):

        self.l =l
        self.b = b


    def area(self):

        return self.l * self.b

    def perimeter(self):

        return 2*(self.l + self.b)

    def is_square(self):
        if self.l == self.b:
            print("This is a square")

        else:

            print("This is not a square")
```

4. **Cylinder Class**

   * Create a `Cylinder` class with `radius` and `height`.
   * Implement:

     * `volume()`
     * `surface_area()`

```python
class Cylinder:

    def __init__(self,r,h):

        self.r =r
        self.h =h

    def volumne(self):

        return 3.14* self.r**2 * self.h

    def curved_surface_area(self):

        return 2 * 3.14 * self.r * self.h





```

```python
c = Cylinder(1,1)
```

```python
c.volumne()
```

```python
c.curved_surface_area()
```
