## Function vs Method
Function: standalone block of code you call by name. It does not require an object/instance.


Method: function attached to an object or class and typically called on that object (can access that object's state).

```python
L = [1,2,3]
L.append(4)
len(L)

'''append is a method while len is a function'''
```

# Creating our own Data type
## Fraction

```python
class Fraction:

    #parameterized constructor : Constructor that needs input
    def __init__(self,x,y):
        self.num=x
        self.den=y

    def __str__(self): # issey pta chalega ki object ko dekhana kaisa hai
        return '{}/{}'.format(self.num,self.den)
    
    def __add__(self,other):
        new_num = self.num* other.den + other.num* self.den
        new_den = self.den*other.den

        return "{}/{}".format(new_num,new_den)
    def __sub__(self,other):
        new_num = self.num* other.den - other.num* self.den
        new_den = self.den*other.den

        return "{}/{}".format(new_num,new_den)
    def __mul__(self,other):
        new_num = self.num* other.num
        new_den = self.den*other.den

        return "{}/{}".format(new_num,new_den)
    def __truediv__(self,other):
        new_num = self.num* other.den
        new_den = self.den*other.num

        return "{}/{}".format(new_num,new_den)
    
    def convert_to_decimal(self):
        return self.num/self.den

        
```

```python
fr1 = Fraction(1,2)
```

```python
print(fr1)
```

```python
fr2 = Fraction(1,2)
```

```python
print(fr2)
```

```python
print(fr1+fr2)
```

```python
fr1.convert_to_decimal()
```

```python
print(fr1-fr2)
```

```python
fr1*fr2
```

```python
fr1/fr2
```
