```python
class Rectangle:

    def __init__(self,l,b):

        self.l =l
        self.b = b


    def area(self):

        return self.l * self.b

    def perimeter(self):

        return 2*(self.l + self.b)

    def is_square(self):
        if self.l == self.b:
            print("This is a square")

        else:

            print("This is not a square")
```

```python
r = Rectangle(12,12)
```

```python
r.area()
```

```python
r.perimeter()
```

```python
r.is_square()
```
